# Enter the user-item rating matrix

```python
user_item_rating = [[1, 2, 5, 0, 1],
                    [5, 0, 1, 2, 5],
                    [0, 3, 4, 3, 4],
                    [1, 1, 5, 4, 0]]  # rows: users, columns: items. 0: missing
print(user_item_rating)
# target user u
target_user = 3   # index from 0
# target item i
target_item = 4   # index from 0
# number of neighbors
k_neighbors = 2
```

# k-NN User-based Collaborative Filtering

Functions: Find common items

```python
def find_common_items(u, v):
  new_u = []
  new_v = []
  for i in range(0, len(u)):
    if u[i] != 0 and v[i]!=0:
        new_u.append(u[i])
        new_v.append(v[i])
  return (new_u, new_v)
```

Functions: Compute average rating

```python
def compute_average_rating(u):
  if (len) == 0:
    return -1
  sum = 0
  n = 0.0
  for r in u:
    if r!=0:
      sum = sum + r
      n = n+1
  return sum / n
```

User-based collaborative filtering

```python
n_users = len(user_item_rating)
n_items = len(user_item_rating[0])
```

```python
# computer average user ratings
avg_ratings = []
for user in user_item_rating:
  avg_ratings.append(compute_average_rating(user))
print(avg_ratings)
```

```python
# compute the predicted score of target item i for target user u
num = 0
denom = 0.00001
for neighbor_key in neighborhood:
  if user_item_rating[neighbor_key][target_item] == 0:
    continue
  num = num + neighborhood[neighbor_key]*(user_item_rating[neighbor_key][target_item]-avg_ratings[neighbor_key])
  denom = denom + abs(neighborhood[neighbor_key])
score = avg_ratings[target_user] + num/denom
print(score)
```

# k-NN Item-based Collaborative Filtering



Functions: Find common users

```python
def find_common_users(i, j):
  new_i = dict()
  new_j = dict()
  for u in range(0, len(i)):
    if i[u] != 0 and j[u]!=0:
        new_i.update({u:i[u]})
        new_j.update({u:j[u]})
  return (new_i, new_j)
```

Functions: Compute average rating

```python
def compute_average_rating(u):
  if (len) == 0:
    return -1
  sum = 0
  n = 0.0
  for r in u:
    if r!=0:
      sum = sum + r
      n = n+1
  return sum / n
```

Item-based collaborative filtering

```python
n_users = len(user_item_rating)
n_items = len(user_item_rating[0])

import numpy as np
temp = np. array(user_item_rating)
temp_transpose = np.transpose(temp)
item_user_rating = temp_transpose.tolist()
print(item_user_rating)
```

```python
# computer average user ratings
avg_ratings = []
for user in user_item_rating:
  avg_ratings.append(compute_average_rating(user))
print(avg_ratings)
```

```python
import math
NaN_val = -9999

# computer similarity between the target item i and every other item j
adjusted_cosine_similarities = dict()
for j in range(0, n_items):
  if j == target_item:
    continue
  common_i, common_j = find_common_users(item_user_rating[target_item], item_user_rating[j])
  std_i = 0
  std_j = 0
  product = 0
  for u in common_i:
    diff_i = common_i[u] - avg_ratings[u]
    diff_j = common_j[u] - avg_ratings[u]
    std_i = std_i + pow(diff_i,2)
    std_j = std_j + pow(diff_j,2)
    product = product + diff_i*diff_j
  #print(j, product,std_i,std_j, common_i, common_j)
  if std_i != 0 and std_j !=0:
    adjusted_cosine_similarities.update({j:product/(math.sqrt(std_i)*math.sqrt(std_j))})
  else:
    adjusted_cosine_similarities.update({j:NaN_num})
print(adjusted_cosine_similarities)
```

```python
# get the neighborhood
from operator import itemgetter
neighborhood = {k: v for k, v in adjusted_cosine_similarities.items() if v > 0}
actual_neighbors = len(neighborhood)
neighborhood = dict(sorted(neighborhood.items(), key = itemgetter(1), reverse = True)[:actual_neighbors])
print(neighborhood)
```

```python
# compute the predicted score of target item i for target user u
num = 0
denom = 0.00001
for neighbor_key in neighborhood:
  num = num + neighborhood[neighbor_key]*user_item_rating[target_user][neighbor_key]
  denom = denom + abs(neighborhood[neighbor_key])
score = num/denom
print(score)
```
